```python
import cv2
import matplotlib as plt
%matplotlib inline
```

```python
img=cv2.imread('tomato.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
print("image matrix shape",img.shape)
print("image matrix shape",img_rgb.shape)
```

```python
print("Data type",img.dtype)
print("Data type",img_rgb.dtype)
```

```python
cv2.imshow('preview', img)
cv2.waitKey(0) # Capital K

cv2.destroyAllWindows() # Capital W, lowercase 'all'
```

```python
cv2.imshow('preview', img_rgb)
cv2.waitKey(0) # Capital K

cv2.destroyAllWindows() # Capital W, lowercase 'all'
```
