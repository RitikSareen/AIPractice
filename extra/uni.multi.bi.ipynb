## Univariate Analysis

```python
df['species'].unique()
# for i in range(len(df['species'].unique())):
#     df_{i} = df[df.loc[df['species']==i]]
```

```python
df_setosa=df.loc[df['species']=='setosa']
df_versicolor=df.loc[df['species']=='versicolor']
df_virginica=df.loc[df['species']=='virginica']
```

```python
df_versicolor
```

```python
plt.plot(df_setosa['sepal_length'],np.ones_like(df_setosa['sepal_length']),'o')
plt.plot(df_versicolor['sepal_length'],np.zeros_like(df_versicolor['sepal_length']),'o')
plt.plot(df_virginica['sepal_length'],np.full_like(df_virginica['sepal_length'],0.5),'o')
```

```python
sns.FacetGrid(df,hue="species", height=4, aspect=1.2).map(plt.scatter,"sepal_length","sepal_width").add_legend();
plt.show()
```
