```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Names":      ["Ritik", "Parth", "Ronit", "Diana", "Sam", "John"],
    "Age":        [26, None, 20, 28, None, 35],
    "Occupation": ["Engineer", "Sales", None, "Engineer", "Sales", None],
    "Salary":     [70000, 50000, None, 80000, None, 60000],
    "Experience": [3, None, 1, None, 2, 10]
})
```

```python
df
```

```python
df=df.fillna({
    'Age': df['Age'].mean(),
    'Salary': df.Salary.median(),
    'Occupation': df.Occupation.fillna("Unknown")
})
df
```
