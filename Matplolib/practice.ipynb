```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.DataFrame({
    "Names":      ["Ritik", "Parth", "Ronit", "Diana", "Sam", "John", "Mia", "Leo"],
    "Department": ["IT", "Sales", "IT", "Sales", "HR", "HR", "IT", "Sales"],
    "Salary_2024": [65000, 48000, 42000, 75000, 40000, 44000, 70000, 50000],
    "Salary_2025": [70000, 50000, 45000, 80000, 43000, 48000, 75000, 52000]
})
```

```python
avg = df.groupby("Department")[["Salary_2024", "Salary_2025"]].mean()
print(avg)
```

```python
departments = avg.index            # ['HR', 'IT', 'Sales']
x = np.arange(len(departments))    # [0, 1, 2] — positions on x-axis
width = 0.35                        # width of each bar

plt.figure(figsize=(8, 5))
plt.bar(x, avg["Salary_2024"], width)

plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.title("Average Salary per Department: 2024 vs 2025")
plt.xticks(x, departments)         # put department names under the pairs
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd

df = pd.DataFrame({
    "Names":      ["Ritik","Parth","Ronit","Diana","Sam","John","Mia","Leo","Zoe","Max"],
    "Department": ["IT","Sales","IT","Sales","HR","HR","IT","Sales","HR","IT"],
    "Age":        [26, 26, 20, 28, 22, 35, 30, 24, 29, 33],
    "Salary":     [70000, 50000, 45000, 80000, 43000, 48000, 75000, 52000, 46000, 68000],
    "Experience": [3, 5, 1, 8, 2, 10, 6, 4, 5, 9],
    "Salary_2024":[65000, 48000, 42000, 75000, 40000, 44000, 70000, 50000, 43000, 64000]
})
```

```python
plt.plot([5, 12, 8, 20, 15])
plt.show()
```

```python
plt.bar(df['Names'],df['Salary'],color='red')
plt.xlabel("Names")
plt.ylabel("Salary")
plt.title("Employee salary")
plt.show()
```

```python
plt.scatter(df['Experience'],df['Salary'],color='red',s=100)
plt.title("Exp v/s Salary")
plt.xlabel("Experience")
plt.ylabel("Salary")
plt.show()
```

```python
import matplotlib.pyplot as plt

df_exp = [1, 5, 1, 8, 2, 10, 6, 4, 5, 9]      # Experience
df_sal = [70000, 50000, 45000, 80000, 43000, 48000, 75000, 52000, 46000, 68000]  # Salary
```

```python
plt.scatter(df_exp, df_sal, color="red", s=100)
plt.xlim(0, 20)        # force x-axis to stretch from 0 to 20
plt.show()
```

```python
plt.scatter(df_exp, df_sal, color="red", s=100)
plt.xlim(4, 8)         # only show experience between 4 and 8
plt.show()
```

```python
plt.scatter(df_exp, df_sal, color="red", s=100)
plt.xticks([0, 5, 10])    # only three labels appear
plt.show()
```
